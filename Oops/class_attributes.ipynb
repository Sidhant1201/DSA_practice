```python
class Circle:
    pi = 3.14 #class var
    
    def __init__(self, radius) -> None:
        self.radius = radius #instance var
```

```python
Circle.pi = 5 #to change class variable you have to access it using class name instead of instance
Circle.pi
```

```python
class RectablgeWithMethod:
    
    def __init__(self, length, breath):
        self.length = length
        self.breath = breath
        
    def calculate_area(self): #instance methods associated to instance through self
        return self.length * self.breath
    
```

```python
rect1 = RectablgeWithMethod(5, 10)
print(rect1.calculate_area())
```
